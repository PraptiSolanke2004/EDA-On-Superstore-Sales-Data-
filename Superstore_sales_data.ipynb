## Data Cleaning 
- Check for missing values 
- Handle missing values (mean, median, mode, or drop)(optional) 
- Check data types and convert if necessary 
- Check for duplicates 

```python
df.isnull().sum()
```

```python
for col in df.columns:
    if df[col].isnull().sum() > 0:
        if df[col].dtype in ["float64", "int64"]:
            df[col] = df[col].fillna(df[col].median())
        else:
            df[col] = df[col].fillna(df[col].mode()[0])
```

## Outlier Detection 
- Use boxplots to identify outliers in Sales, Profit, and Discount

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["Sales"],color='Blue')

plt.title("Boxplot of Sales")
plt.xlabel("Sales")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["Profit"])

plt.title("Boxplot of Profit")
plt.xlabel("Profit")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["Discount"])

plt.title("Boxplot of Discount")
plt.xlabel("Discount")

plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(data=df[["Sales", "Profit", "Discount"]])

plt.title("Outlier Detection: Sales, Profit and Discount")
plt.xlabel("Variables")
plt.ylabel("Values")

plt.show()
```

```python
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

sales_outliers = df[
    (df["Sales"] < lower_bound) |
    (df["Sales"] > upper_bound)
]

print("Number of Sales Outliers:", len(sales_outliers))
```

```python
Q1 = df["Profit"].quantile(0.25)
Q3 = df["Profit"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

profit_outliers = df[
    (df["Profit"] < lower_bound) |
    (df["Profit"] > upper_bound)
]

print("Number of Profit Outliers:", len(profit_outliers))
```

```python
Q1 = df["Discount"].quantile(0.25)
Q3 = df["Discount"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

discount_outliers = df[
    (df["Discount"] < lower_bound) |
    (df["Discount"] > upper_bound)
]

print("Number of Discount Outliers:", len(discount_outliers))
```
